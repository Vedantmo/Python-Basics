2.What is an interpreter in Python?

An interpreter in Python is a special program that reads and runs your code line by line. It checks each line, understands it, and then executes it right away. This means you don’t need to compile the entire code before running it like in some other languages. If there's an error, Python stops at that line and shows the problem. The interpreter makes Python easier to test and debug, which is great for beginners. It also allows you to run code quickly and see results instantly in tools like the Python shell or IDLE.



3.What are pre-defined keywords in Python?

Pre-defined keywords in Python are special reserved words that have specific meanings in the language. You can’t use them for variable names or anything else. Examples include if, else, for, while, True, False, and None. These keywords tell Python how to run the code or describe certain actions. They are part of the Python syntax and help the interpreter understand the structure and logic of your program. Since their meaning is fixed, using them for other purposes like naming variables would confuse the interpreter and cause errors.



4.Can keywords be used as variable names?

No, you cannot use keywords as variable names in Python. Keywords are reserved words that already have a special meaning in the language. If you try to use them as variable names, Python will get confused and show an error. For example, using if = 5 or while = "hello" will not work because if and while are keywords used for decision-making and loops. Instead, you should pick names that are not in the list of keywords. You can check Python's keywords by using the keyword module with keyword kwlist.


5.What is mutability in Python?

Mutability in Python means whether a data type can be changed after it is created. If a value can be changed, it’s called mutable. If it cannot be changed, it’s called immutable. For example, lists are mutable because you can add, remove, or change items in them. But strings and tuples are immutable, meaning their values cannot be changed once set. Understanding mutability helps you decide how to store and work with data. It also affects memory usage and performance in your program when working with large sets of data.

6.Why are lists mutable, but tuples are immutable?

Lists are mutable because they are meant to store items that may change—like adding, removing, or updating elements. This makes lists useful when you need flexibility in your data. Tuples, on the other hand, are immutable because they are designed for data that shouldn't be changed. This makes them faster and safer to use, especially when passing around fixed data. Also, immutable types like tuples can be used as dictionary keys, while mutable ones like lists cannot. Python gives you both options so you can choose based on your needs: changeable (list) or fixed (tuple).

7.What is the difference between “==” and “is” operators in Python?

In Python, == checks if two values are equal, while is checks if they are the exact same object in memory. For example, a == b returns True if the values of a and b are the same. But a is b only returns True if both a and b point to the same object. You might have two lists with the same content, so == would be True, but is would be False because they are stored separately. Use == for value comparison and is for identity checks.

8.What are logical operators in Python?

Logical operators in Python are used to combine or modify conditions. The three main ones are and, or, and not. The and operator checks if both conditions are true. The or operator checks if at least one condition is true. The not operator reverses the result—if something is true, not makes it false. These operators help in making decisions in programs. For example, if age > 18 and has_id: checks if someone is an adult and has ID before allowing entry. They’re very useful in conditional statements and loops

9.What is type casting in Python?

Type casting in Python means changing the data type of a value from one type to another. For example, you can convert a string like "123" into an integer using int("123"). This is useful when you get input as a string but want to do math with it. Python supports many conversions like int(), float(), str(), and bool(). Type casting helps you make sure the values are in the correct format for your operations. It prevents errors and allows you to work with different types of data together.

10.What is the difference between implicit and explicit type casting?

Implicit type casting happens automatically when Python converts one type to another without you asking. For example, if you add an integer to a float (3 + 2.5), Python turns the result into a float (5.5). Explicit type casting is when you tell Python to change the type using functions like int(), float(), or str(). You do this when you want more control over how data is used or displayed. Implicit is automatic and safe, while explicit is manual and gives you more control, but you must use it correctly to avoid errors.

11.What is the purpose of conditional statements in Python?

Conditional statements in Python help your program make decisions based on certain conditions. The most common ones are if, elif, and else. They let you run specific blocks of code only when a condition is true. For example, you can check a person's age, and if it's over 18, allow access to a website. If not, show a warning. Without conditional statements, your program would run everything no matter what. With them, you can make your code smarter and more flexible, like a real decision-maker.

12.How does the elif statement work?

The elif statement in Python stands for "else if." It's used when you want to check multiple conditions in an if block. First, Python checks the if condition. If that’s false, it moves to elif, and so on. If one elif condition is true, it runs that block and skips the rest. If none are true, it runs the final else (if present). This helps avoid writing many separate if statements and keeps your code clean. It’s useful for choices like grading systems or menu options where several conditions might apply.

13.What is the difference between for and while loops?

A for loop is used when you know how many times you want to repeat something—like going through a list or counting numbers. A while loop runs as long as a condition is true, even if you don’t know how many times that will be. Use for when you’re iterating over items, and while when the ending depends on a changing condition. For example, use for to print numbers 1 to 10, and while to keep asking a user for input until they type “exit.” Both are useful for different tasks.

14.Describe a scenario where a while loop is more suitable than a for loop.

A while loop is better when you don’t know in advance how many times the loop should run. For example, imagine making a login system that keeps asking the user for a password until they get it right. You don’t know how many tries they’ll need, so a while loop works perfectly. It will keep running as long as the password is wrong. This kind of loop is great for real-time checks, repeated user input, or conditions that depend on user actions or random results—not fixed counts.

6.Write a function to demonstrate the behavior of mutable and immutable arguments.

```python
def modify_arguments(immutable_arg, mutable_arg):
    print("\nInside function before modification:")
    print("Immutable:", immutable_arg)
    print("Mutable:", mutable_arg)

    immutable_arg += 100
    mutable_arg.append(100)

    print("\nInside function after modification:")
    print("Immutable:", immutable_arg)
    print("Mutable:", mutable_arg)

num = 50                # Immutable (int)
my_list = [1, 2, 3]     # Mutable (list)

print("Before function call:")
print("Immutable:", num)
print("Mutable:", my_list)

modify_arguments(num, my_list)

print("\nAfter function call:")
print("Immutable:", num)         # Still 50 (unchanged)
print("Mutable:", my_list)       # List is modified
```

7.Write a program that performs basic arithmetic operations on two user-input numbers.

9.Write a Python program to convert user input from string to integer, float, and boolean types.

```python

user_input = input("Enter a value: ")

try:
    int_value = int(user_input)
    print(f"Integer conversion: {int_value}")
except ValueError:
    print("Cannot convert to integer.")

try:
    float_value = float(user_input)
    print(f"Float conversion: {float_value}")
except ValueError:
    print("Cannot convert to float.")

bool_value = bool(user_input)
print(f"Boolean conversion: {bool_value}")
```

10.Write code to demonstrate type casting with list elements.

```python

str_list = ["10", "20", "30", "40.5", "50"]


int_list = []
for item in str_list:
    try:
        int_list.append(int(item))
    except ValueError:
        print(f"Cannot convert '{item}' to int.")
        int_list.append(None)


float_list = []
for item in str_list:
    try:
        float_list.append(float(item))
    except ValueError:
        print(f"Cannot convert '{item}' to float.")
        float_list.append(None)


print("\nOriginal string list:", str_list)
print("Converted to integers:", int_list)
print("Converted to floats:  ", float_list)
```

14.Write a program to reverse a string using a while loop.

```python
text = input("Enter a string to reverse: ")
reversed_str = ""
index = len(text) - 1

while index >= 0:
    reversed_str += text[index]
    index -= 1

print("Reversed string:", reversed_str)
```

15.Write a Python program to calculate the factorial of a number provided by the user using a while loop.

```python
num = int(input("Enter a number to calculate its factorial: "))
factorial = 1
n = num

if num < 0:
    print("Factorial does not exist for negative numbers.")
else:
    while n > 1:
        factorial *= n
        n -= 1
    print(f"Factorial of {num} is {factorial}")
```
